# Sales Analysis — Q1 2024

This notebook analyses sales data from Q1 2024.
We look at revenue by region, top products, and month-on-month trends.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('sales_q1.csv')
print(df.head())
```

```python
total = df['revenue'].sum()
top_region = df.groupby('region')['revenue'].sum().idxmax()
print(f'Total revenue: £{total:,.0f}')
print(f'Top region: {top_region}')
```

## Key Findings

- Revenue up 18% vs Q4 2023
- North region outperformed all others
- Widget Pro drove 31% of total revenue
- March was the strongest month

```python
monthly = df.groupby('month')['revenue'].sum()
monthly.plot(kind='bar', title='Monthly Revenue Q1 2024')
plt.tight_layout()
plt.savefig('monthly_revenue.png')
```
